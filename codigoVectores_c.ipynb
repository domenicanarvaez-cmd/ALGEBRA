```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import date

# Define the vectors and scalar from the last executed code segment
vector1 = np.array([-2, 5])
vector2 = np.array([3, 4])
scalar = -4
scalar_new = 0.5

# Calculate scalar multiplication (scalar * vector2)
scaled_vector_v2 = scalar * vector2

# Calculate new scalar multiplication (0.5 * vector1)
scaled_vector_v1 = scalar_new * vector1

# Calculate vector sum (vector1 + vector2)
sum_vector = vector1 + vector2

# Create the plot
fig, ax = plt.subplots(figsize=(10, 10))

# Plot vector1
ax.quiver(0, 0, vector1[0], vector1[1], angles='xy', scale_units='xy', scale=1, color='red', label=f'Vector 1 ({vector1})', width=0.007)

# Plot vector2
ax.quiver(0, 0, vector2[0], vector2[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'Vector 2 ({vector2})', width=0.007)

# Plot scaled_vector_v2
ax.quiver(0, 0, scaled_vector_v2[0], scaled_vector_v2[1], angles='xy', scale_units='xy', scale=1, color='purple', label=f'Scaled Vector ({scalar} * Vector 2)', width=0.009)

# Plot scaled_vector_v1
ax.quiver(0, 0, scaled_vector_v1[0], scaled_vector_v1[1], angles='xy', scale_units='xy', scale=1, color='orange', label=f'Scaled Vector ({scalar_new} * Vector 1)', width=0.009)

# Plot sum_vector
ax.quiver(0, 0, sum_vector[0], sum_vector[1], angles='xy', scale_units='xy', scale=1, color='green', label=f'Sum Vector ({vector1} + {vector2})', width=0.009)

# Set limits dynamically based on all vector components
all_coords = np.concatenate([vector1, vector2, scaled_vector_v2, scaled_vector_v1, sum_vector])
max_abs_coord = np.max(np.abs(all_coords))

ax.set_xlim([-max_abs_coord - 2, max_abs_coord + 2])
ax.set_ylim([-max_abs_coord - 2, max_abs_coord + 2])
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.6)
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('Scalar Multiplication and Vector Sum in 2D Plane')
ax.legend()

# Show the plot
plt.show()

print(f"Original Vector 1: {vector1}")
print(f"Original Vector 2: {vector2}")
print(f"Scalar: {scalar}")
print(f"Scaled Vector ({scalar} * vector2): {scaled_vector_v2}")
print(f"Scalar_new: {scalar_new}")
print(f"Scaled Vector ({scalar_new} * vector1): {scaled_vector_v1}")
print(f"Sum of Vectors (vector1 + vector2): {sum_vector}")

# Personal Data
today = date.today()
nombres = "Domenica Brigeeth Narvaez Rivas" # Using the variable from kernel state
carrera = "Ingienería en Computación" # Using the variable from kernel state
ciclo = "1° 'A'" # Placeholder for the user to fill

print()
print(f"Datos:")
print(f"Nombre: {nombres}")
print(f"Fecha: {today}")
print(f"Carrera: {carrera}")
print(f"Ciclo: {ciclo}")
```
